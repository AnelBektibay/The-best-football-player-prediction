## RANDOM FOREST REGRESSOR

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor()
```

```python
rf_params = {
    'n_jobs': -1,
    'n_estimators': 500,
     'warm_start': True, 
    'max_depth': 6,
    'min_samples_leaf': 2,
    'max_features' : 'sqrt',
    'verbose': 0
}
```

```python
rf.fit(X_train, y_train)
```

```python
randf_pred=rf.predict(X_test)
```

```python
from sklearn import metrics
print("mean_squared_error (mse) = " + str(metrics.mean_squared_error(y_test, randf_pred)))
print("rmse = " + str(np.sqrt(metrics.mean_squared_error(y_test, randf_pred))))
print("r2_score (r2) = " + str(metrics.r2_score(y_test, randf_pred)))
print("mean_absolute_error (mae) = " + str(metrics.mean_absolute_error(y_test, randf_pred)))
```

## LIGHTGBM

```python
import lightgbm as lgb
```

```python
hyper_params = {
    'task': 'train',
    'boosting_type': 'gbdt',
    'objective': 'regression',
    'metric': ['l2', 'auc'],
    'learning_rate': 0.005,
    'feature_fraction': 0.9,
    'bagging_fraction': 0.7,
    'bagging_freq': 10,
    'verbose': 0,
    "max_depth": 4,
    "num_leaves": 128,  
    "max_bin": 512,
    "num_iterations": 100000,
    "n_estimators": 10
}
```

```python
lgbm = lgb.LGBMRegressor()
```

```python
lgbm.fit(X_train, y_train,
        eval_set=[(X_test, y_test)],
        eval_metric='mse',
        early_stopping_rounds=1000)
```

```python
lgbm_pred = lgbm.predict(X_test, num_iteration=lgbm.best_iteration_)
```

```python
print("mean_squared_error (mse) = " + str(metrics.mean_squared_error(y_test, lgbm_pred)))
print("rmse = " + str(np.sqrt(metrics.mean_squared_error(y_test, lgbm_pred))))
print("r2_score (r2) = " + str(metrics.r2_score(y_test, lgbm_pred)))
print("mean_absolute_error (mae) = " + str(metrics.mean_absolute_error(y_test, lgbm_pred)))
```
